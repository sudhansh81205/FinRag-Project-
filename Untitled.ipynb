```python
import os
from bs4 import BeautifulSoup

# 1. Locate the downloaded files (now including .txt)
filing_paths = []
for root, dirs, files in os.walk("sec-edgar-filings"):
    for file in files:
        if file.endswith((".html", ".htm", ".txt")):
            path = os.path.join(root, file)
            filing_paths.append(path)

# Safeguard to check if files were found before accessing index 0
if not filing_paths:
    print("0 documents found. Ensure the 'sec-edgar-filings' folder is in the exact same directory as this Jupyter notebook.")
else:
    print(f"Found {len(filing_paths)} primary documents. Loading the first one:")
    print(filing_paths[0])
    
    # 2. Parse the HTML/text of the first document
    with open(filing_paths[0], "r", encoding="utf-8") as file:
        soup = BeautifulSoup(file.read(), "lxml")

    # 3. Extract the clean narrative text
    text_content = soup.get_text(separator=' ', strip=True)
    print("\n--- Extracted Text (First 500 chars) ---")
    print(text_content[:500])

    # 4. Isolate the financial tables for your multimodal pipeline
    tables = soup.find_all("table")
    print(f"\nFound {len(tables)} tables in this document.")



```

```python
import pandas as pd
from io import StringIO
from langchain_text_splitters import RecursiveCharacterTextSplitter

# --- 1. CONVERTING HTML TABLES TO MARKDOWN ---
markdown_tables = []

# Loop through the raw HTML tables found by BeautifulSoup
for table in tables:
    try:
        # Convert the table to a string, then wrap it in StringIO to fix the warning
        html_string = str(table)
        df = pd.read_html(StringIO(html_string))[0] 
        
        # Convert the data grid into a clean, text-based Markdown format
        md_table = df.to_markdown(index=False)
        markdown_tables.append(md_table)
    except ValueError:
        # Skip any invisible layout tables that don't contain actual data
        continue

print(f"Successfully converted {len(markdown_tables)} tables to Markdown.")
if markdown_tables:
    print("\n--- Example Markdown Table (First 200 chars) ---")
    print(markdown_tables[0][:200]) 


# --- 2. CHUNKING THE NARRATIVE TEXT ---
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,     
    chunk_overlap=200,   
    separators=["\n\n", "\n", ".", " ", ""] 
)

text_chunks = text_splitter.split_text(text_content)

print(f"\nSuccessfully split the narrative into {len(text_chunks)} manageable chunks.")
```

```python
import chromadb
from sentence_transformers import SentenceTransformer

# 1. Initialize a persistent Vector Database on your local hard drive
# This ensures your stored data is retained after you close the notebook
chroma_client = chromadb.PersistentClient(path="./sec_vector_db")

# 2. Create a "collection" (which acts just like a table in a standard SQL database)
collection = chroma_client.get_or_create_collection(name="sec_10k_filings")

# 3. Load the open-source embedding model 
# 'all-MiniLM-L6-v2' is fast, efficient, and standard for text semantic search
embedding_model = SentenceTransformer('all-MiniLM-L6-v2')

# 4. For this initial test, we will only embed the first 50 chunks of your narrative text
test_chunks = text_chunks[:50]

# Convert the English text chunks into mathematical arrays (vectors)
print("Generating embeddings... this may take a moment.")
embeddings = embedding_model.encode(test_chunks).tolist()

# 5. Create unique IDs and metadata tags for each chunk
ids = [f"chunk_{i}" for i in range(len(test_chunks))]
metadatas = [{"source": "narrative_text"} for _ in range(len(test_chunks))]

# 6. Store the IDs, the vectors, the original text, and the metadata into ChromaDB
collection.add(
    ids=ids,
    embeddings=embeddings,
    documents=test_chunks,
    metadatas=metadatas
)

print(f"Success! Your database now contains {collection.count()} vectors.")
```

```python
# 1. The user asks a question
user_question = "What are the company's biggest financial risks?"

# 2. We use the exact same Sentence Transformer to turn the question into numbers
question_vector = embedding_model.encode(user_question).tolist()

# 3. We ask ChromaDB to find the 2 closest matching text chunks
results = collection.query(
    query_embeddings=[question_vector],
    n_results=2 # Return the top 2 matches
)

print(f"Question: {user_question}\n")
print("--- Top Result Found by ChromaDB ---")
print(results['documents'][0][0])
```
