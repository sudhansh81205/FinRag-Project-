```python
import pandas as pd
from io import StringIO
from langchain_text_splitters import RecursiveCharacterTextSplitter

# --- 1. CONVERTING HTML TABLES TO MARKDOWN ---
markdown_tables = []

# Loop through the raw HTML tables found by BeautifulSoup
for table in tables:
    try:
        # Convert the table to a string, then wrap it in StringIO to fix the warning
        html_string = str(table)
        df = pd.read_html(StringIO(html_string))[0] 
        
        # Convert the data grid into a clean, text-based Markdown format
        md_table = df.to_markdown(index=False)
        markdown_tables.append(md_table)
    except ValueError:
        # Skip any invisible layout tables that don't contain actual data
        continue

print(f"Successfully converted {len(markdown_tables)} tables to Markdown.")
if markdown_tables:
    print("\n--- Example Markdown Table (First 200 chars) ---")
    print(markdown_tables[0][:200]) 


# --- 2. CHUNKING THE NARRATIVE TEXT ---
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,     
    chunk_overlap=200,   
    separators=["\n\n", "\n", ".", " ", ""] 
)

text_chunks = text_splitter.split_text(text_content)

print(f"\nSuccessfully split the narrative into {len(text_chunks)} manageable chunks.")
```

```python
import chromadb
from sentence_transformers import SentenceTransformer

# 1. Initialize a persistent Vector Database on your local hard drive
# This ensures your stored data is retained after you close the notebook
chroma_client = chromadb.PersistentClient(path="./sec_vector_db")

# 2. Create a "collection" (which acts just like a table in a standard SQL database)
collection = chroma_client.get_or_create_collection(name="sec_10k_filings")

# 3. Load the open-source embedding model 
# 'all-MiniLM-L6-v2' is fast, efficient, and standard for text semantic search
embedding_model = SentenceTransformer('all-MiniLM-L6-v2')

# 4. For this initial test, we will only embed the first 50 chunks of your narrative text
test_chunks = text_chunks[:50]

# Convert the English text chunks into mathematical arrays (vectors)
print("Generating embeddings... this may take a moment.")
embeddings = embedding_model.encode(test_chunks).tolist()

# 5. Create unique IDs and metadata tags for each chunk
ids = [f"chunk_{i}" for i in range(len(test_chunks))]
metadatas = [{"source": "narrative_text"} for _ in range(len(test_chunks))]

# 6. Store the IDs, the vectors, the original text, and the metadata into ChromaDB
collection.add(
    ids=ids,
    embeddings=embeddings,
    documents=test_chunks,
    metadatas=metadatas
)

print(f"Success! Your database now contains {collection.count()} vectors.")
```

```python
import os
import pandas as pd
from io import StringIO
from bs4 import BeautifulSoup
from langchain_text_splitters import RecursiveCharacterTextSplitter
import chromadb
from sentence_transformers import SentenceTransformer

print("1. Loading and parsing the SEC document...")
filing_paths = []
for root, dirs, files in os.walk("sec-edgar-filings"):
    for file in files:
        if file.endswith((".html", ".htm", ".txt")):
            path = os.path.join(root, file)
            filing_paths.append(path)

# Parse the first document
with open(filing_paths[0], "r", encoding="utf-8") as file:
    soup = BeautifulSoup(file.read(), "lxml")

# Extract Narrative Text
text_content = soup.get_text(separator=' ', strip=True)

# Extract Tables
tables = soup.find_all("table")

print("2. Chunking text and converting tables to Markdown...")
# Chunk the text
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
text_chunks = text_splitter.split_text(text_content)

# Convert tables
markdown_tables = []
for table in tables:
    try:
        df = pd.read_html(StringIO(str(table)))[0] 
        markdown_tables.append(df.to_markdown(index=False))
    except ValueError:
        continue

print("3. Filtering data for the database...")
# Skip tiny fragments and grab a meaningful slice of the data
filtered_text_chunks = [c for c in text_chunks if len(c.strip()) > 80][:300]
filtered_tables = [t for t in markdown_tables if len(t.strip()) > 50][:100]

combined_docs = filtered_text_chunks + filtered_tables
combined_metadatas = (
    [{"type": "narrative_text"} for _ in filtered_text_chunks] +
    [{"type": "financial_table"} for _ in filtered_tables]
)
combined_ids = [f"doc_{i}" for i in range(len(combined_docs))]

print(f"4. Generating embeddings for {len(combined_docs)} items (This will take a minute)...")
embedding_model = SentenceTransformer('all-MiniLM-L6-v2')
doc_embeddings = embedding_model.encode(combined_docs, show_progress_bar=True).tolist()

print("5. Saving to Vector Database...")
chroma_client = chromadb.PersistentClient(path="./sec_vector_db")
collection = chroma_client.get_or_create_collection(name="sec_10k_filings")

# Upsert (update or insert) prevents duplicate errors if you run this cell twice
collection.upsert(
    ids=combined_ids,
    embeddings=doc_embeddings,
    documents=combined_docs,
    metadatas=combined_metadatas
)

print(f"Success! Index updated. Total documents in vector DB: {collection.count()}")
```

```python
cleaned_tables = []

for table in tables:
    try:
        # 1. Grab the text immediately above the table to act as its title/context
        prev_tag = table.find_previous(["p", "div", "h1", "h2", "h3", "h4"])
        table_context = prev_tag.get_text(separator=" ", strip=True)[:250] if prev_tag else "Financial Table"
        
        # 2. Read the HTML table into Pandas
        df = pd.read_html(StringIO(str(table)))[0]
        
        # 3. Remove completely empty rows and columns, then replace remaining 'nan' with blanks
        df = df.dropna(how="all", axis=0).dropna(how="all", axis=1)
        df = df.fillna("")
        
        # 4. Skip tiny layout tables (require at least 2 rows and 2 columns)
        if df.shape[0] < 2 or df.shape[1] < 2:
            continue
            
        # 5. Glue the context heading to the top of the Markdown table
        md_table = f"Table Context: {table_context}\n\n" + df.to_markdown(index=False)
        cleaned_tables.append(md_table)
        
    except ValueError:
        continue

print(f"Cleaned {len(cleaned_tables)} valid financial tables.")
print("\n--- Example Cleaned Table with Context ---")
print(cleaned_tables[10][:500]) # Print a sample table from inside the report
```

```python
# Select 150 cleaned tables to embed
selected_tables = cleaned_tables[:150]
table_ids = [f"clean_table_{i}" for i in range(len(selected_tables))]
table_metadatas = [{"type": "clean_financial_table"} for _ in selected_tables]

print("Embedding cleaned tables...")
table_embeddings = embedding_model.encode(selected_tables, show_progress_bar=True).tolist()

# Save to ChromaDB
collection.upsert(
    ids=table_ids,
    embeddings=table_embeddings,
    documents=selected_tables,
    metadatas=table_metadatas
)

# Test the search on the newly cleaned tables
query_vector = embedding_model.encode(user_question).tolist()
clean_table_results = collection.query(
    query_embeddings=[query_vector],
    n_results=2,
    where={"type": "clean_financial_table"}
)

print("\n=== NEW RETRIEVED CLEAN TABLES ===")
for i, t in enumerate(clean_table_results['documents'][0]):
    print(f"\n[Clean Table {i+1}]:\n{t[:450]}...")
```

```python
import json

def run_comparison_experiment(question, ground_truth_hint):
    # 1. Encode the user question into a vector
    q_vec = embedding_model.encode(question).tolist()
    
    # 2. Retrieve Narrative Text (Used by BOTH Baseline and FinRAG)
    text_res = collection.query(
        query_embeddings=[q_vec],
        n_results=2,
        where={"type": "narrative_text"}
    )
    narrative_context = "\n\n".join(text_res['documents'][0])
    
    # 3. Retrieve Cleaned Tables (Used ONLY by FinRAG)
    table_res = collection.query(
        query_embeddings=[q_vec],
        n_results=2,
        where={"type": "clean_financial_table"}
    )
    table_context = "\n\n".join(table_res['documents'][0])[:4000]
    
    # 4. Generate Standard Baseline RAG Answer (Text-Only)
    baseline_prompt = f"""
    You are a financial analyst. Answer the question using ONLY the narrative text below.
    If the exact numbers are missing, state what is available.
    
    --- NARRATIVE EVIDENCE ---
    {narrative_context}
    
    --- QUESTION ---
    {question}
    """
    baseline_ans = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=baseline_prompt
    ).text
    
    # 5. Generate FinRAG Answer (Narrative Text + Cleaned Tables)
    finrag_prompt = f"""
    You are a financial analyst. Answer the question using BOTH the narrative text and financial tables below.
    Provide exact dollar figures and years.
    
    --- NARRATIVE EVIDENCE ---
    {narrative_context}
    
    --- TABULAR EVIDENCE ---
    {table_context}
    
    --- QUESTION ---
    {question}
    """
    finrag_ans = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=finrag_prompt
    ).text
    
    # 6. Automated Evaluation (Measure Accuracy and Faithfulness from 1 to 10)
    eval_prompt = f"""
    You are an objective academic evaluator grading two RAG systems on a scale of 1 to 10.
    
    Question: {question}
    Expected Ground Truth: {ground_truth_hint}
    
    --- SYSTEM A (Standard Text-Only Baseline) ---
    {baseline_ans}
    
    --- SYSTEM B (FinRAG Multimodal Hybrid) ---
    {finrag_ans}
    
    Provide a brief score comparison formatted as:
    - Baseline Accuracy Score (1-10):
    - Baseline Faithfulness Score (1-10):
    - FinRAG Accuracy Score (1-10):
    - FinRAG Faithfulness Score (1-10):
    - Reason for score difference:
    """
    eval_report = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=eval_prompt
    ).text
    
    print("=== 1. STANDARD BASELINE RAG (TEXT-ONLY) ===")
    print(baseline_ans)
    print("\n=== 2. FINRAG (TEXT + CLEANED TABLES) ===")
    print(finrag_ans)
    print("\n=== 3. AUTOMATED EVALUATION METRICS ===")
    print(eval_report)

# Run the experiment on your revenue question
run_comparison_experiment(
    question="What are the company's total net sales and breakdown between Products and Services for 2021, 2020, and 2019?",
    ground_truth_hint="Total net sales: $365,817M in 2021 ($297,392M Products, $68,425M Services), $274,515M in 2020, $260,174M in 2019."
)
```
